```python
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
nlp.pipe_names
```

```python
doc = nlp('Ansh is founder of google')
```

```python
doc.text
```

```python
doc.ents
```

```python
for ent in doc.ents:
    print(ent,'--> ',spacy.explain(ent.label_).split(',')[0])
```

```python
from spacy import displacy
```

```python
text = displacy.render(doc,style='ent')
```
