```python
temp = string.punctuation
```

```python
temp
```

```python
def remove_punctuation(doc):
    for i in temp:
        if i in doc:
            doc = doc.replace(i,' ')
    return doc.strip()
```

```python
remove_punctuation('my name is ansh,ansh i boy!./')
```

```python
temp = stopwords.words('english')
```

```python
from nltk.tokenize import word_tokenize

```

```python
temp
```

```python
def remove_stopwords(text):
    tokens = word_tokenize(text)
    filtered_text = [word for word in tokens if word.lower() not in temp]
    return ' '.join(filtered_text)
```

```python
print(remove_stopwords('My name is ansh'))
```

```python
temp = df.review.apply(remove_stopwords)
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
ps.stem("walked")
```

```python
def stem_text(text):
    stemed_list=[]
    tokens = word_tokenize(text)
    for i in tokens:
        stemed_list.append(ps.stem(i))
    return ' '.join(stemed_list)
```

```python
stem_text('no one expects the star trek movies to be high expects star trek movies high art fans')
```

```python
remove_stopwords('no one expects the star trek movies to be high expects star trek movies high art fans')
```

```python
df.text_processed = df.text_processed.apply(stem_text)
```
